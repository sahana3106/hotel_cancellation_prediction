```python
print('numerical:',list(numeric_features))
print('categorial:',list(categorical_features))
```

```python
categorical_features = categorical_features.tolist()
categorical_features.append("agent")

numeric_features = numeric_features.drop("agent")
```

```python
print('numerical:',list(numeric_features))
print('categorial:',categorical_features)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler,OneHotEncoder,FunctionTransformer
```

```python
numeric_transformer=Pipeline(steps=[
    ("imputer",SimpleImputer(strategy='median')),
    ('scaler',StandardScaler())
])
```

```python
def to_string(x):
    return x.astype(str)
```

```python
categorical_transformer=Pipeline(steps=[
    ('imputer',SimpleImputer(strategy='constant',fill_value='unknown')),
    ('to_string',FunctionTransformer(to_string
                                    )),
    ('onehot',OneHotEncoder(handle_unknown='ignore'))
])
```

```python

preprocessor=ColumnTransformer(
            transformers=[
                ('num',numeric_transformer,numeric_features),
                ('cat',categorical_transformer,categorical_features)
                
            ]
)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_pipeline=Pipeline(steps=[
    ('preprocessor',preprocessor),
    ('model',LogisticRegression(max_iter=1000))
])
```

```python
logistic_pipeline.fit(x_train,y_train)
```

```python
y_pred=logistic_pipeline.predict(x_test)
```

```python
y_prob = logistic_pipeline.predict_proba(x_test)[:, 1]
```

```python
from sklearn.metrics import classification_report,confusion_matrix,ConfusionMatrixDisplay
```

```python
print(classification_report(y_test,y_pred))
```

```python
cm=confusion_matrix(y_test,y_pred)
disp=ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=['not_canceled','canceled']
)
disp.plot()
plt.show()
```

```python
from sklearn.metrics import roc_auc_score,roc_curve
y_prob=logistic_pipeline.predict_proba(x_test)[:,1]
auc=roc_auc_score(y_test,y_prob)
print(auc)
```

```python
result=[]
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
```

```python
def evaluate_model(name,y_test,y_pred,y_prob=None):
    result={
        'model':name,
        'accuracy':accuracy_score(y_test,y_pred),
         'precision':precision_score(y_test,y_pred),
         'recall':recall_score(y_test,y_pred),
          'f1_score':f1_score(y_test,y_pred)
    }
    if y_pred is not None:
        result['roc_auc']=roc_auc_score(y_test,y_pred)
    else:
        result['roc_auc']=None
    return result    
```

```python
results = []

results.append(
    evaluate_model("Logistic Regression", y_test, y_pred, y_prob)
)

results.append(
    evaluate_model("Random Forest", y_test, rf_pred, rf_prob)
)

results.append(
    evaluate_model("XGBoost", y_test, xgb_pred, xg_prob)
)

results.append(
    evaluate_model("SVM", y_test, svm_pred)
)
```

```python
results_df = pd.DataFrame(results)

results_df
```
